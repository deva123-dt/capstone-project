```python
conversation_history = []

def ask_question(query):
    global conversation_history

    results = vectorstore.similarity_search(
        query,
        k=3
    )

    context = "\n\n".join(
        f"Paper: {doc.metadata['paper_title']}\n"
        f"Page: {doc.metadata['page_number']}\n"
        f"Content: {doc.page_content}"
        for doc in results
    )

    history = "\n".join(
        f"User: {q}\nAssistant: {a}"
        for q, a in conversation_history[-3:]
    )

    prompt = f"""
You are a research paper question-answering assistant.

Previous conversation:
{history}

Current question:
{query}

Research Paper Context:
{context}

Answer using ONLY the research paper context.
If the answer is not available, say:
"I could not find the answer in the provided research papers."
"""

    response = llm.invoke(prompt)

    conversation_history.append(
        (query, response.content)
    )

    print("Question:", query)
    print("\nAnswer:", response.content)


ask_question("What is LoRA?")
```
