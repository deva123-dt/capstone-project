```python
import os
from langchain_community.document_loaders import PyPDFLoader

PDF_FOLDER = "data/papers"

TITLE_MAP = {
    "lora.pdf": "LoRA: Low-Rank Adaptation of Large Language Models",
    "qlora.pdf": "QLoRA: Efficient Finetuning of Quantized LLMs",
    "rag.pdf": "Retrieval-Augmented Generation for Knowledge-Intensive NLP Tasks",
    "researchpaper 1.pdf": "Attention Is All You Need"
}

documents = []

for file in os.listdir(PDF_FOLDER):
    if file.endswith(".pdf"):
        path = os.path.join(PDF_FOLDER, file)
        loader = PyPDFLoader(path)
        pages = loader.load()

        for page in pages:
            page.metadata["paper_title"] = TITLE_MAP.get(
                file,
                os.path.splitext(file)[0]
            )
            page.metadata["page_number"] = page.metadata.get("page", 0) + 1

        documents.extend(pages)

print("Total pages loaded:", len(documents))
```

```python
for doc in documents[:5]:
    print("Paper:", doc.metadata["paper_title"])
    print("Page:", doc.metadata["page_number"])
    print("Text:", doc.page_content[:150])
    print("-" * 60)
    
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_documents(documents)

print("Total pages:", len(documents))
print("Total chunks:", len(chunks))
```

```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(
    model="nomic-embed-text"
)

query = "What is LoRA in large language models?"

vector = embeddings.embed_query(query)

print("Embedding created successfully!")
print("Embedding dimensions:", len(vector))
print("First 10 values:", vector[:10])
```

```python
from langchain_chroma import Chroma

vectorstore = Chroma(
    collection_name="research_papers",
    embedding_function=embeddings,
    persist_directory="vectorstore"
)

print("Vector database connected successfully!")
print("Total chunks in vector database:", vectorstore._collection.count())
```

```python
query = "What is LoRA and how does it reduce trainable parameters?"

results = vectorstore.similarity_search(
    query,
    k=3
)

print("Query:", query)
print("\nTop 3 Retrieved Sources:\n")

for i, doc in enumerate(results, 1):
    print(f"{i}. {doc.metadata['paper_title']}")
    print(f"   Page: {doc.metadata['page_number']}")
    print(f"   Content: {doc.page_content[:300]}")
    print("-" * 70)
```

```python
retriever = vectorstore.as_retriever(
    search_type="mmr",
    search_kwargs={
        "k": 3,
        "fetch_k": 10
    }
)

mmr_results = retriever.invoke(
    "What is LoRA and how does it reduce trainable parameters?"
)

print("MMR Retrieval Results:\n")

for i, doc in enumerate(mmr_results, 1):
    print(f"{i}. {doc.metadata['paper_title']}")
    print(f"   Page: {doc.metadata['page_number']}")
    print(f"   Content: {doc.page_content[:300]}")
    print("-" * 70)
```

```python
from rank_bm25 import BM25Okapi

query = "What is LoRA and how does it reduce trainable parameters?"

dense_results = vectorstore.similarity_search(
    query,
    k=3
)

corpus = [
    doc.page_content
    for doc in chunks
]

tokenized_corpus = [
    text.lower().split()
    for text in corpus
]

bm25 = BM25Okapi(tokenized_corpus)

tokenized_query = query.lower().split()

scores = bm25.get_scores(tokenized_query)

top_indices = sorted(
    range(len(scores)),
    key=lambda i: scores[i],
    reverse=True
)[:3]

bm25_results = [
    chunks[i]
    for i in top_indices
]

combined_results = dense_results + bm25_results

unique_results = []
seen = set()

for doc in combined_results:

    key = (
        doc.metadata["paper_title"],
        doc.metadata["page_number"],
        doc.page_content[:100]
    )

    if key not in seen:
        seen.add(key)
        unique_results.append(doc)

hybrid_results = unique_results[:3]

print("Hybrid Search Results:\n")

for i, doc in enumerate(hybrid_results, 1):
    print(f"{i}. {doc.metadata['paper_title']}")
    print(f"   Page: {doc.metadata['page_number']}")
    print(f"   Content: {doc.page_content[:300]}")
    print("-" * 70)
```

```python
import numpy as np

query = "What is LoRA and how does it reduce trainable parameters?"

query_vector = np.array(
    embeddings.embed_query(query)
)

scored_results = []

for doc in hybrid_results:

    doc_vector = np.array(
        embeddings.embed_query(doc.page_content)
    )

    similarity = np.dot(
        query_vector,
        doc_vector
    ) / (
        np.linalg.norm(query_vector) *
        np.linalg.norm(doc_vector)
    )

    scored_results.append(
        (doc, similarity)
    )

scored_results.sort(
    key=lambda x: x[1],
    reverse=True
)

reranked_results = [
    doc
    for doc, score in scored_results
]

print("Cosine Similarity Reranking:\n")

for i, (doc, score) in enumerate(scored_results, 1):
    print(f"{i}. {doc.metadata['paper_title']}")
    print(f"   Page: {doc.metadata['page_number']}")
    print(f"   Similarity Score: {score:.4f}")
    print("-" * 70)
```

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="gemma3:1b",
    temperature=0
)

query = "What is LoRA and how does it reduce trainable parameters?"

context = "\n\n".join(
    f"Paper: {doc.metadata['paper_title']}\n"
    f"Page: {doc.metadata['page_number']}\n"
    f"Content: {doc.page_content}"
    for doc in reranked_results[:3]
)

prompt = f"""
You are a research paper question-answering assistant.

Answer the question using ONLY the research paper context provided below.

Question:
{query}

Research Paper Context:
{context}

Rules:
1. Use only the provided research paper context.
2. Do not add information from your own knowledge.
3. If the answer is not available in the context, say:
"I could not find the answer in the provided research papers."
4. Give a clear and concise answer.
"""

response = llm.invoke(prompt)

print("Question:")
print(query)

print("\nAnswer:")
print(response.content)
```

```python
test_questions = [
    "What is LoRA and how does it reduce trainable parameters?",
    "What is the main idea behind the Transformer architecture?",
    "What is Retrieval-Augmented Generation?",
    "How does RAG improve knowledge-intensive NLP tasks?",
    "What is QLoRA?",
    "How does QLoRA reduce memory requirements?",
    "What is the purpose of attention in the Transformer?",
    "What are the advantages of low-rank adaptation?",
    "How does LoRA differ from full fine-tuning?",
    "What problem does Retrieval-Augmented Generation solve?"
]

print("Total test questions:", len(test_questions))

for i, question in enumerate(test_questions, 1):
    print(f"{i}. {question}")
```

```python
evaluation_results = []

for number, query in enumerate(test_questions, 1):

    results = vectorstore.similarity_search(
        query,
        k=3
    )

    context = "\n\n".join(
        f"Paper: {doc.metadata['paper_title']}\n"
        f"Page: {doc.metadata['page_number']}\n"
        f"Content: {doc.page_content}"
        for doc in results
    )

    prompt = f"""
You are a research paper question-answering assistant.

Answer the question using ONLY the research paper context.

Question:
{query}

Context:
{context}

Rules:
1. Use only the provided context.
2. Do not use outside knowledge.
3. If the answer is not available, say:
"I could not find the answer in the provided research papers."
4. Give a clear and concise answer.
"""

    response = llm.invoke(prompt)

    sources = " | ".join(
        f"{doc.metadata['paper_title']} - Page {doc.metadata['page_number']}"
        for doc in results
    )

    evaluation_results.append({
        "question": query,
        "answer": response.content,
        "top_3_sources": sources
    })

    print(f"\nQUESTION {number}")
    print(query)
    print("\nANSWER")
    print(response.content)
    print("\nSOURCES")
    print(sources)
    print("=" * 80)
```

```python
import pandas as pd

evaluation_df = pd.DataFrame(evaluation_results)

evaluation_df.to_csv(
    "experiments/notebook_evaluation_results.csv",
    index=False,
    encoding="utf-8"
)

print("Evaluation results saved successfully!")
print("Total questions evaluated:", len(evaluation_df))
```
